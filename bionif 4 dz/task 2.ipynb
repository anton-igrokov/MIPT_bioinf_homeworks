```python
from Bio import motifs
from Bio import SeqIO
from Bio.Seq import Seq

# motif 
sites = ["GAGGTAAAC", "TCCGTAAGC", "CAGGTTGGA", "ACAGTCAGC", "TAGGTCAGC",
         "CAGGTCAGC", "CAGGTCGAT", "CAGGTCAGC", "CAGGTCAGC", "CAGGTTGGC"]

sites = [Seq(s) for s in sites]
motif = motifs.create(sites)

pwm = motif.counts.normalize(pseudocounts=0.1).log_odds()
print(motif.consensus)
```

```python
for record in SeqIO.parse("/Users/antonkorotkov/Homo_sapiens.GRCh38.dna.chromosome.1.fa", "fasta"):
    record = record[:10**6]
    print("ID:", record.id)
    print("Description:", record.description)
    print("Length:", len(record.seq))
    print("First 100 bp:", record.seq[:100])

```

```python
# поиск по прямой цепи
search_results = motif.pssm.search(record, threshold=0.5)

# перенесем из хитрого обьекта в список
hits_forward = []
for position, score in search_results:
    hits_forward.append((position, score))

print(f"Количество сайтов связывания на прямой цепи: {len(hits_forward)}")
# первые 10 хитов
for pos, sc in hits_forward[:10]:  
    print(f"Позиция: {pos}, Скор: {sc:.2f}")

# делаем реверс цепь и аналогично
record_reverse = record.reverse_complement()
search_reverse = motif.pssm.search(record_reverse, threshold=0.5)


hits_reverse = []
for position, score in search_reverse:
    hits_reverse.append((position, score))

print(f"\nКоличество сайтов связывания на обратной цепи: {len(hits_reverse)}")
for pos, sc in hits_reverse[:10]:
    print(f"Позиция: {pos}, Скор: {sc:.2f}")
```

```python
output_file = "motif_hits.txt"

with open(output_file, "w") as f:
    f.write("Цепь\tПозиция\tСкор\n")
    
    for pos, score in hits_forward:
        f.write(f"Прямая\t{pos}\t{score:.2f}\n")
    
    for pos, score in hits_reverse:
        f.write(f"Обратная\t{pos}\t{score:.2f}\n")
```
