```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p = np.array([[0.3553719 , 0.01652893, 0.23553719, 0.39256198],
       [0.66964286, 0.01785714, 0.17857143, 0.13392857],
       [0.15909091, 0.08181818, 0.25      , 0.50909091],
       [0.10823529, 0.20470588, 0.20705882, 0.48      ]])
matrices = []
for i in range(1, 1001):
    matrices.append(np.linalg.matrix_power(p, i))
    if i % 100 == 0 or i < 10:
        print(matrices[i-1])
```

```python
# в общем после условно 10 ниче почти не меняется
norm = []
for i in range(1000 - 1):
    norm.append(np.linalg.norm(matrices[i+1] - matrices[i], 'fro'))
plt.plot(range(1, 1000), norm)
```

```python
plt.plot(range(1, 10), norm[:9])
```
