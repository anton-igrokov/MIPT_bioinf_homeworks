```python
from Bio import SeqIO
import matplotlib.pyplot as plt
import numpy as np
```

```python
modern_human = "/Users/antonkorotkov/homework3/yung.fasta"
neander_human = "/Users/antonkorotkov/homework3/neander.fasta"
for record in SeqIO.parse(modern_human, "fasta"):
    print("ID:", record.id)
    print("Description:", record.description)
    print("Sequence:", str(record.seq))
    modern_human = str(record.seq)


for record in SeqIO.parse(neander_human, "fasta"):
    print("ID:", record.id)
    print("Description:", record.description)
    print("Sequence:", str(record.seq))
    neander_human = str(record.seq)
```

```python
# параметры использовал как на семинаре. скользим окном 20
n, m = len(modern_human), len(neander_human)
dp = [[0 for i in range(n - 20 + 1)] for j in range(m - 20 + 1)]
for i in range(n - 20 + 1):
    for j in range(m - 20 + 1):
        if modern_human[i: i + 20] == neander_human[j: j + 20]:
            dp[j][i] = 1
```

```python
# Get coordinates where value == 1
massive = np.array(dp)
rows, cols = np.where(massive == 1)

# Plot
plt.figure(figsize=(6, 6))
plt.scatter(cols, rows, s=1)  
plt.gca().invert_yaxis()      
plt.show()
```
