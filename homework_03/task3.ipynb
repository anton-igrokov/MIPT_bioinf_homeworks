```python
def needlman(seq1: str, seq2: str,
             match: int, mismatch: int, gap: int):
    """все числа неотрицательные на вход (ну если вы штрафовать хотите а не награждать за мисметч)"""
    # сначала инициализируем все
    n = len(seq1)
    m = len(seq2)
    dp = [[None for _ in range(n+1)] for _ in range(m+1)]
    
    # заполняем нулевые строки гэпами
    for i in range(n+1):
        dp[0][i] = - gap * i
        
    for j in range(m+1):
        dp[j][0] = - gap * j

    # теперь реализуем алгоритм динамического прогр.
    for i in range(1, n+1):
        for j in range(1, m + 1):
            score = match if seq1[i-1] == seq2[j-1] else -mismatch

            dp[j][i] = max(
                dp[j-1][i] - gap,
                dp[j][i-1] - gap,
                dp[j-1][i-1] + score
            )
    return dp



needlman("ATGCAGCAGCAGCCA", "ATATAT", 1, 1, 4)
        
```

```python
import math

def needleman_wunsch_affine_score(seq1, seq2, match=1, mismatch=-1,
                                  gap_open=-10, gap_extend=-1):
    # инициализация
    n = len(seq1)
    m = len(seq2)
    
    NEG_INF = -math.inf

    M  = [[NEG_INF]*(m+1) for _ in range(n+1)]
    Ix = [[NEG_INF]*(m+1) for _ in range(n+1)]      # вспомогательная матрица для гэпов по оси х (т е в сек1)
    Iy = [[NEG_INF]*(m+1) for _ in range(n+1)]      # вспомогательная матрица для гэпов по оси у (т е в сек2)

    M[0][0] = 0

    # инициализируем первые строки
    for i in range(1, n+1):
        Ix[i][0] = gap_open + (i-1)*gap_extend

    for j in range(1, m+1):
        Iy[0][j] = gap_open + (j-1)*gap_extend

    def s(a, b):
        return match if a == b else mismatch

    for i in range(1, n+1):
        for j in range(1, m+1):

            M[i][j] = max(
                M[i-1][j-1],
                Ix[i-1][j-1],
                Iy[i-1][j-1]
            ) + s(seq1[i-1], seq2[j-1])    # матрица М НЕ содержит гэпы. только метч или мисматч

            Ix[i][j] = max(                # гепы тут во вспом матрицах, поэтому если переходим из м - откртие
                M[i-1][j] + gap_open,
                Ix[i-1][j] + gap_extend
            )

            Iy[i][j] = max(
                M[i][j-1] + gap_open,
                Iy[i][j-1] + gap_extend
            )

    return max(M[n][m], Ix[n][m], Iy[n][m])


needleman_wunsch_affine_score("ATGCAGCAGCAGCCA", "ATATAT")
```
